# CantoNLU Sentiment Analysis Initial EDA

```python
## 1. Load Dataset

import pandas as pd

train = pd.read_json("../data/sentiment/train.jsonl", lines=True)
valid = pd.read_json("../data/sentiment/valid.jsonl", lines=True)
test = pd.read_json("../data/sentiment/test.jsonl", lines=True)
```

```python
## 2. Dataset Dimensions

print(train.shape)
print(valid.shape)
print(test.shape)
```

```python
## 3. Dataset Structure

print(train.columns)
print(valid.columns)
print(test.columns)

print(train.head())
print(valid.head())
print(test.head())
```

```python
## 4. Sample Input Reviews

for i in range(5):
    print("Review", train.iloc[i]["sentence"])
    print("Label:", train.iloc[i]["label"])
    print()

for i in range(5):
    print("Review", valid.iloc[i]["sentence"])
    print("Label:", valid.iloc[i]["label"])
    print()

for i in range(5):
    print("Review", test.iloc[i]["sentence"])
    print("Label:", test.iloc[i]["label"])
    print()



```

```python
## 5. Expected Outputs/Labels

print("Labels:")
print(train["label"].unique())

print("\nLabel counts:")
print(train["label"].value_counts())

print("\nMissing labels:")
print(train["label"].isnull().sum())

print("\nSample reviews and labels:")

print("Train")
for i in range(5):
    print("Review:", train.iloc[i]["sentence"])
    print("Label:", train.iloc[i]["label"])
    print()

print(" \n Valid")
for i in range(5):
    print("Review:", valid.iloc[i]["sentence"])
    print("Label:", valid.iloc[i]["label"])
    print()

print("\n Test")
for i in range(5):
    print("Review:", test.iloc[i]["sentence"])
    print("Label:", test.iloc[i]["label"])
    print()

for label in train["label"].unique():

    print("\nLabel:", label)

    examples = train[train["label"] == label].sample(3)

    for _, row in examples.iterrows():
        print("Review:", row["sentence"])
```

```python
## 6. Label Distribution

print("Train")
print(train["label"].value_counts())
print(train["label"].value_counts(normalize=True) * 100)

print("\nValidation")
print(valid["label"].value_counts())
print(valid["label"].value_counts(normalize=True) * 100)

print("\nTest")
print(test["label"].value_counts())
print(test["label"].value_counts(normalize=True) * 100)
```

```python
## 7. Missing Values

print("Train")
train.isna().sum()

print("\n Validation")
valid.isna().sum()

print("\n Test")
test.isna().sum()
```

```python
## 8. Duplicate Reviews

print("Train")
train.duplicated().sum()

print("\n Validation")
valid.duplicated().sum()

print("\n Test")
test.duplicated().sum()
```

```python
## 9. Review Input Length

train["text_length"] = train["sentence"].str.len()
valid["text_length"] = valid["sentence"].str.len()
test["text_length"] = test["sentence"].str.len()

print("TRAINING")
print(train["text_length"].describe())

print("\nVALID")
print(valid["text_length"].describe())

print("\nTEST")
print(test["text_length"].describe())
```

```python
## 10. Limitations

# Label distribution
print("Label distribution:")
print(train["label"].value_counts())

print("\nLabel percentages:")
print(train["label"].value_counts(normalize=True) * 100)

# Train/Test overlap
train_test_overlap = set(train["sentence"]) & set(test["sentence"])

print("\nTrain/Test overlapping reviews:", len(train_test_overlap))
```

```python
## 11. Unusual Data

# Very short reviews
short_reviews = train[train["text_length"] <= 5]

print("Number of very short reviews:", len(short_reviews))
print(short_reviews[["sentence", "label", "text_length"]].head(10))


# Reviews containing repeated punctuation
repeated_punctuation = train[
    train["sentence"].str.contains(
        r"[!！?？。]{2,}",
        regex=True,
        na=False
    )
]

print("\nReviews with repeated punctuation:", len(repeated_punctuation))
print(repeated_punctuation[["sentence", "label"]].head(10))
```
